```python
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
data = np.load("operation_blackout_encoded (1).npy")

# Inspect dataset
print("Original Shape:", data.shape)
print("Total Elements:", data.size)

# Reshape
matrix = data.reshape(12, 12)

print("\nReshaped Matrix:")
print(matrix)
```

```python
plt.figure(figsize=(7, 7))
plt.imshow(matrix, cmap="binary", interpolation="nearest")
plt.title("Operation Blackout - Original Matrix")
plt.colorbar()
plt.show()
```

```python

# STEP 3: Matrix Transformations

transposed = matrix.T
horizontal_flip = matrix[:, ::-1]
vertical_flip = matrix[::-1, :]
reverse_both = matrix[::-1, ::-1]
transposed_reverse = matrix.T[::-1, :]

print("Original:", matrix.shape)
print("Transposed:", transposed.shape)
print("Horizontal Flip:", horizontal_flip.shape)
print("Vertical Flip:", vertical_flip.shape)
print("Reverse Both:", reverse_both.shape)
print("Transposed + Reverse:", transposed_reverse.shape)
```

```python

fig, axes = plt.subplots(2, 3, figsize=(13, 9))

images = [
    (matrix, "1. Original"),
    (transposed, "2. Transposed"),
    (horizontal_flip, "3. Horizontal Flip"),
    (vertical_flip, "4. Vertical Flip"),
    (reverse_both, "5. Reverse Both"),
    (transposed_reverse, "6. Transposed + Reverse")
]

for ax, (img, title) in zip(axes.flat, images):
    ax.imshow(img, cmap="binary", interpolation="nearest")
    ax.set_title(title)
    ax.axis("off")

plt.suptitle("OPERATION BLACKOUT - Transformation Analysis")
plt.tight_layout()
plt.show()
```

```python

# STEP 4: Invert binary pixel values

inverted_matrix = 1 - matrix

print("Original Matrix:")
print(matrix)

print("\nInverted Matrix:")
print(inverted_matrix)
```

```python

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(matrix, cmap="binary", interpolation="nearest")
axes[0].set_title("Original Matrix")

axes[1].imshow(inverted_matrix, cmap="binary", interpolation="nearest")
axes[1].set_title("Inverted Matrix")

for ax in axes:
    ax.axis("off")

plt.suptitle("Operation Blackout - Pixel Inversion")
plt.tight_layout()
plt.show()
```

```python

# STEP 5: Combined transformations

combinations = [
    (matrix, "Original"),
    (matrix.T, "Transpose"),
    (matrix[::-1, :], "Vertical Reverse"),
    (matrix[:, ::-1], "Horizontal Reverse"),
    (matrix.T[::-1, :], "Transpose + Reverse"),
    (1 - matrix, "Inverted"),
    (1 - matrix.T, "Inverted Transpose"),
    (1 - matrix[::-1, ::-1], "Inverted + Reverse")
]

fig, axes = plt.subplots(2, 4, figsize=(14, 8))

for ax, (img, title) in zip(axes.flat, combinations):
    ax.imshow(img, cmap="binary", interpolation="nearest")
    ax.set_title(title)
    ax.axis("off")

plt.suptitle("Operation Blackout - Combined Transformation Analysis")
plt.tight_layout()
plt.show()
```

```python

# STEP 6: Detailed pixel-level visualization

fig, axes = plt.subplots(1, 4, figsize=(16, 5))

rotations = [
    (matrix, "Original"),
    (np.rot90(matrix, 1), "90 Degrees"),
    (np.rot90(matrix, 2), "180 Degrees"),
    (np.rot90(matrix, 3), "270 Degrees")
]

for ax, (img, title) in zip(axes, rotations):
    ax.imshow(img, cmap="binary", interpolation="nearest")
    ax.set_title(title)
    ax.set_xticks(np.arange(-0.5, 12, 1))
    ax.set_yticks(np.arange(-0.5, 12, 1))
    ax.grid(color="gray", linewidth=0.5)
    ax.set_xticklabels([])
    ax.set_yticklabels([])

plt.suptitle("Operation Blackout - Pixel Orientation Analysis")
plt.tight_layout()
plt.show()
```

```python
 
# STEP 7: Binary Pattern Analysis

print("Number of black pixels (1s):", np.count_nonzero(matrix))
print("Number of white pixels (0s):", np.count_nonzero(matrix == 0))

print("\nBlack pixels in each row:")
print(np.sum(matrix, axis=1))

print("\nBlack pixels in each column:")
print(np.sum(matrix, axis=0))

# Check horizontal and vertical symmetry
horizontal_symmetry = np.array_equal(matrix, matrix[::-1, :])
vertical_symmetry = np.array_equal(matrix, matrix[:, ::-1])

print("\nHorizontal Symmetry:", horizontal_symmetry)
print("Vertical Symmetry:", vertical_symmetry)
```

```python

# STEP 8: Final pattern investigation

final_matrix = np.rot90(matrix, 3)

plt.figure(figsize=(9, 9))

plt.imshow(
    final_matrix,
    cmap="binary",
    interpolation="nearest"
)

plt.title("Operation Blackout - Final Decoding View")

plt.xticks(np.arange(12))
plt.yticks(np.arange(12))
plt.grid(color="gray", linewidth=0.5)

plt.show()

print("Final Matrix:")
print(final_matrix)
```

```python
# Compare all four rotations with inverted colors

fig, axes = plt.subplots(2, 4, figsize=(14, 8))

for i in range(4):
    rotated = np.rot90(matrix, i)

    axes[0, i].imshow(rotated, cmap="binary",
                      interpolation="nearest")
    axes[0, i].set_title(f"Rotation {i * 90}°")
    axes[0, i].axis("off")

    axes[1, i].imshow(1 - rotated, cmap="binary",
                      interpolation="nearest")
    axes[1, i].set_title(f"Inverted {i * 90}°")
    axes[1, i].axis("off")

plt.suptitle("Operation Blackout - Final Orientation Comparison")
plt.tight_layout()
plt.show()
```

```python
# Save the final visualization

plt.figure(figsize=(8, 8))

plt.imshow(np.rot90(matrix, 3), cmap="binary",
           interpolation="nearest")

plt.title("Operation Blackout - Final Result")
plt.axis("off")

plt.savefig("operation_blackout_final.png",
            dpi=300, bbox_inches="tight")

plt.show()

print("Final visualization saved successfully!")
print("Original array shape:", data.shape)
print("Final matrix shape:", final_matrix.shape)
print("Total pixels:", data.size)
```

```python
# FINAL DECODING

decoded_matrix = np.rot90(matrix, 3)

plt.figure(figsize=(8, 8))

plt.imshow(decoded_matrix, cmap="binary",
           interpolation="nearest")

plt.title("Operation Blackout - Decoded Character: Q",
          fontsize=16)

plt.axis("off")

plt.savefig(
    "operation_blackout_decoded.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Decoded Character: Q")
print("Transformation: 270-degree rotation")
print("Final Shape:", decoded_matrix.shape)
```
